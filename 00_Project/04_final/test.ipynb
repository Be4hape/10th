```python
import pandas as pd
import plotly.express as px

df_phase = df_tele[df_tele["_T"] == "LogPhaseChange"].copy()

# 예시: ts(시간), phase, 그리고 metric 한 개가 있다고 가정
fig = px.scatter_3d(
    df_phase,
    x="ts",          # 또는 gameTime, frame 등
    y="phase",
    z="metric",      # 예: gold, damage, etc.
    color="win",     # 있으면 (0/1)
    opacity=0.6
)
fig.show()
```
